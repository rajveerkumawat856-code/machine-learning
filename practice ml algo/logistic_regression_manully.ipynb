```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv('heart_disease_cleaned.csv')
```

```python
df.head()
```

```python
def disease(value):
    if value==0:
        return 'No'
    else:
        return 'Yes'
```

```python
df['disease']=df['num'].apply(disease)
```

```python
df.head()
```

```python
df.isnull().sum()
```

```python
df=df.drop(columns=['num','id'])
```

```python
df.head()
```

```python
df.columns
```

```python
df.isnull().sum()
```

```python
numerical_data = ['age', 'trestbps', 'chol', 'thalch', 'oldpeak','ca']
categorical_data = ['sex', 'dataset', 'cp', 'fbs', 'restecg','exang', 'thal']
ordinal_data = ['slope']
```

```python
from sklearn.preprocessing import OneHotEncoder,OrdinalEncoder,LabelEncoder,StandardScaler
from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer

```

```python
df['slope'].unique()
```

```python
tranform=ColumnTransformer(transformers=[('nominal',OneHotEncoder(handle_unknown='ignore'),categorical_data),('scaling',StandardScaler(),numerical_data),('ordinal',OrdinalEncoder(categories=[['downsloping','flat','upsloping']]),ordinal_data)])
```

```python
X=df[['age', 'sex', 'dataset', 'cp', 'trestbps', 'chol', 'fbs', 'restecg',
       'thalch', 'exang', 'oldpeak', 'slope', 'ca', 'thal']]
```

```python
Y=df['disease']
```

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
```

```python
X=tranform.fit_transform(X)
```

```python
le=LabelEncoder()
Y=le.fit_transform(Y)
```

```python
X_train,X_test,Y_train,Y_test=train_test_split(X,Y,test_size=0.2)
```

```python
from sklearn.linear_model import LogisticRegression
```

```python
lr=LogisticRegression()
```

```python
lr.fit(X_train,Y_train)
```

```python
y_predict=lr.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score,r2_score
```

```python

accuracy_score(Y_test,y_predict)*100
```

```python
r2_score(Y_test,y_predict)
```
