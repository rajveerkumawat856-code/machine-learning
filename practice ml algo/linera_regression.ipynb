# now we will apply the feature encoding

```python
df.columns
```

```python
nominal_data=['city','country','State']
numerical_data=[ 'bedrooms', 'bathrooms', 'sqft_living', 'sqft_lot', 'floors',
       'waterfront', 'view', 'condition', 'sqft_above', 'sqft_basement',
       'yr_built', 'yr_renovated', 'Month', 'Day', 'Year',
        'Zip']
```

```python
from sklearn.preprocessing import OneHotEncoder
from sklearn.preprocessing import StandardScaler
oh=OneHotEncoder()
ss=StandardScaler()
```

```python
from sklearn.compose import ColumnTransformer
```

```python
transnform=ColumnTransformer(transformers=[('nominal',OneHotEncoder(handle_unknown='ignore'),nominal_data),('scaling',StandardScaler(),numerical_data)])
```

```python
df.columns
```

```python
X=df[['bedrooms', 'bathrooms', 'sqft_living', 'sqft_lot', 'floors',
       'waterfront', 'view', 'condition', 'sqft_above', 'sqft_basement',
       'yr_built', 'yr_renovated', 'city', 'country', 'Month', 'Day', 'Year',
       'State', 'Zip']]
```

```python
Y=df['price']
```

```python
X
```

```python
Y
```

```python
X=transnform.fit_transform(X)
```

```python
X
```

```python
Y
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,Y_train,Y_test=train_test_split(X,Y,test_size=0.2)
```

```python
X_train
```

```python
from sklearn.linear_model import LinearRegression
lr=LinearRegression()
```

```python
lr.fit(X_train,Y_train)
```

```python
y_predict=lr.predict(X_test)
```

```python
y_predict
```

```python
from sklearn.metrics import mean_absolute_error
mean_absolute_error(Y_test,y_predict)
```

```python
from sklearn.metrics import r2_score
r2_score(Y_test,y_predict)
```
