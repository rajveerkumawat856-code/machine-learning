# converter ---> bring the transform or change specific values in a col with another value & we use this function remember it is a function

```python
# define a funcation

def station(value):
    if value=='Q':
        return 'Queenstown'
    else:
        return value
```

```python
pd.read_csv('Titanic.csv',converters={'Embarked':station})
```
