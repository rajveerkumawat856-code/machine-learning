Chatbot

Responds to input, mostly in a single turn or a simple back-and-forth conversation
Follows a fairly fixed pattern: user asks → model responds
No real ability to take actions in the world — it just generates text
Doesn't typically maintain complex, multi-step goals; it reacts rather than plans
Example: a customer support bot that answers FAQs based on a knowledge base

AI Agent

Given a goal, and it figures out the steps needed to achieve it
Can use tools (search the web, call APIs, run code, query a database, write files) to gather information or take real actions
Has a loop: plan → act → observe result → re-plan if needed, repeating until the goal is done or it decides to stop
Maintains state/context across multiple steps, not just multiple conversational turns
Example: an agent that's told "book me the cheapest flight to Mumbai next Friday" — it searches flights, compares prices, and actually completes the booking, adjusting its approach if something fails (e.g., a site is down, so it tries another)
