```python
df.head()
```

# univeriate analysis on numerical col
## -->Data distrubtion of numerical col --> describe() & plotting ---> kdeplot & histogram
## --> outliear detection ---> using box plot


```python
df['Age'].describe()
```

```python
df['Age'].median()
```

```python
df[df['Age']==76.0]
```

```python
sns.boxplot(data=df,x='Age')
plt.show()
```

```python
df['Age'].isnull().sum()
```

```python
df['Age']=df['Age'].fillna(df['Age'].median())
```

```python
df['Age'].isnull().sum()
```

```python
sns.histplot(data=df,x='Age')
plt.grid()
plt.show()
```

# categorical categorical --> crosstab() --> two categories as input on which we can perform the analysis

```python
x=pd.crosstab(df['Pclass'],df['Survived'],normalize=True)*100

# by normalize we get in percentage 
```

```python
sns.heatmap(x)
```

```python
x=pd.crosstab(df['Sex'],df['Survived'],normalize=True)*100
```

```python
sns.heatmap(x)
```
