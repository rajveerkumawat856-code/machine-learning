# feature encoding 
#### Feature encoding is the process of converting data (especially categorical or text data) into a numerical format that machine learning algorithms can understand.

## ordinal data-->
### Ordinal data is a type of categorical data where the categories have a meaningful order or ranking, but the difference between the categories is not necessarily equal.

## EX:
| Size        | Order |
| ----------- | ----- |
| Small       | 1st   |
| Medium      | 2nd   |
| Large       | 3rd   |
| Extra Large | 4th   |


# Nominal data --->
## it is a type of categorical data where the categories have no natural order or ranking

## ex:

| Color  |
| ------ |
| Red    |
| Blue   |
| Green  |
| Yellow |
