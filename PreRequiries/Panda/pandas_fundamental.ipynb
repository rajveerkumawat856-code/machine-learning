# pandas series
#### What is a Series in Pandas?

A Pandas Series is a one-dimensional labeled array that can store data of a single type (such as integers, floats, strings, or booleans). Each value has an associated index.

Think of a Series as a single column of a table.

```python
pd.Series(data=[40,50,60],index=['bread','milk','butter'],name='price of gorcery items') 
```

# dataframe
In Pandas, a DataFrame is a two-dimensional table-like data structure used to store and manipulate data.

Think of it like an Excel spreadsheet or a SQL table:

| Name    | Age | City     |
| ------- | --- | -------- |
| Alice   | 25  | New York |
| Bob     | 30  | London   |
| Charlie | 35  | Paris    |

Rows represent records (observations).
Columns represent variables (features).
Each column can contain different data types (numbers, text, dates, etc.).
