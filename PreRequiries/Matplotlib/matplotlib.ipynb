```python
m = m.groupby('Date')['Amount'].sum()

month_order = [
    'January', 'February', 'March', 'April',
    'May', 'June', 'July', 'August'
]

m.index = pd.CategoricalIndex(
    m.index,
    categories=month_order,
    ordered=True
)

m = m.sort_index()

```

```python
n = n.groupby('Date')['Amount'].sum()

month_order = [
    'January', 'February', 'March', 'April',
    'May', 'June', 'July', 'August'
]

n.index = pd.CategoricalIndex(
    n.index,
    categories=month_order,
    ordered=True
)

n = n.sort_index()
```

```python
n
```

```python
o = o.groupby('Date')['Amount'].sum()

month_order = [
    'January', 'February', 'March', 'April',
    'May', 'June', 'July', 'August'
]

o.index = pd.CategoricalIndex(
    o.index,
    categories=month_order,
    ordered=True
)

o = o.sort_index()
```

# if some way if we convert the labeling of bar on x axis to numeric value the we can seperate the bar so we will do this with the help of numpy(np.arange())

```python
plt.figure(figsize=(7,4))
plt.title('Product based Total boxes shipped')
plt.xlabel('Month Name')
plt.ylabel('Total Sales')
plt.bar(np.arange(m.shape[0])-0.2,m.values,color='green',label='Mint Chips Choco',width=0.2)
plt.bar(np.arange(n.shape[0]),n.values,color='red',label='85% Dark Bars',width=0.2)
plt.bar(np.arange(o.shape[0])+0.2,o.values,color='yellow',label='Peanut Butter Cubes',width=0.2)
plt.legend()
plt.show()
```

# stacked bar chart
#### instead of keeping the bar side by side direction,we can keep the bar in the staked manner i.e,one on top of other and in order to figure out which category is having max sale,we can have the look over the lenght of the bar


#### stacked bar chart --> bar() ---> parameter --> bottom --> this bottom parameter specify that what bar i want to keep below my current bar

#### red bar
#### green bar-->bottom --> red bar
#### blue bar -->bottom --> green bar +red bar

### as we want to differenciate on the basis of bar lenght value so that's why we have to keep y_value of previous bar at the bottom

```python
plt.Figure(figsize=(7,4))
plt.title('Product based Total boxes shipped')
plt.xlabel('Month Name')
plt.ylabel('Total Sales')
plt.bar(m.index,m.values,color='green',label='Mint Chips Choco')
plt.bar(n.index,n.values,color='red',label='85% Dark Bars',bottom=m.values)
plt.bar(o.index,o.values,color='yellow',label='Peanut Butter Cubes',bottom=(m.values+n.values))
plt.xticks(np.arange(m.shape[0])-0.2,x)
plt.legend()
plt.show()
```

# histogram

#### A Histogram in Matplotlib is a graph that shows the distribution (frequency) of numerical data. It tells you how many values fall within a certain range (bin).

#### Why do we use a Histogram?
 To understand the distribution of data.

To identify the most common value range.

To detect outliers.

To check if data is normal, skewed, or uniform.

# pie chart

#### pie chart is a circular graph used to show the proportion or percentage of different categories in a dataset. Each slice represents a part of the whole.

Why do we use a Pie Chart?

To show percentages.

To compare parts of a whole.

To visualize category proportions.

Best when there are only a few categories (3–6).
