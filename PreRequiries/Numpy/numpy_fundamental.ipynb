# numpy
#### NumPy is an open-source Python library used for numerical computing. It provides the ndarray, a homogeneous, multi-dimensional array that is more memory-efficient and faster than Python lists. NumPy also offers built-in functions for mathematical operations, linear algebra, statistics, random number generation, and array manipulation. It is widely used in data science, machine learning, deep learning, and scientific computing.

#### At the core of numpy package is th ndimentional array object.

# why numpy?

#### 1. Speed: Operations on large arrays are much faster than using Python lists because NumPy is implemented largely in C.
#### 2. Arrays: It provides the ndarray object, which stores homogeneous data efficiently.
#### 3. Mathematics: Supports vectorized operations, linear algebra, statistics, Fourier transforms, and more.
#### 4. Memory efficiency: Uses less memory than equivalent Python data structures.
#### 5. Scientific ecosystem: Libraries like Pandas, SciPy, and TensorFlow build on NumPy arrays.

# What is a NumPy Array?
#### - A NumPy array (ndarray) is a fast, efficient data structure for numerical data.
##### - Stores elements of the same data type.
#### - Supports multi-dimensional arrays (1D, 2D, 3D, ...).
#### - Optimized for mathematical and scientific computations.


#### 1. Array Creation Functions
- np.array()       -> Create an array from a list or tuple.
- np.zeros()       -> Create an array filled with zeros.
- np.ones()        -> Create an array filled with ones.
- np.empty()       -> Create an empty array.
- np.arange()      -> Create an array with evenly spaced values.
- np.linspace()    -> Create an array with evenly spaced values over an interval.
- np.eye()         -> Create an identity matrix.
- np.full()        -> Create an array filled with a specified value.

#### 2. Array Information Functions
- arr.ndim         -> Number of dimensions.
- arr.shape        -> Shape of the array.
- arr.size         -> Total number of elements.
- arr.dtype        -> Data type of the array.
- arr.itemsize     -> Size of each element in bytes.

#### 3. Array Conversion Functions
- astype()         -> Change the data type of an array.
- tolist()         -> Convert a NumPy array to a Python list.

#### 4. Array Reshaping Functions
- reshape()        -> Change the shape of an array.
- flatten()        -> Convert to a 1D array (copy).
- ravel()          -> Convert to a 1D array (view when possible).
- transpose()      -> Transpose an array.
- T                -> Shortcut for transpose.

#### 5. Joining and Splitting Functions
- concatenate()    -> Join arrays.
- stack()          -> Stack arrays along a new axis.
- hstack()         -> Stack horizontally.
- vstack()         -> Stack vertically.
- split()          -> Split an array.
- hsplit()         -> Split horizontally.
- vsplit()         -> Split vertically.

#### 6. Mathematical Functions
- add()            -> Addition.
- subtract()       -> Subtraction.
- multiply()       -> Multiplication.
- divide()         -> Division.
- power()          -> Raise to a power.
- sqrt()           -> Square root.
- abs()            -> Absolute value.

#### 7. Statistical Functions
- sum()            -> Sum of elements.
- mean()           -> Average.
- median()         -> Median.
- std()            -> Standard deviation.
- var()            -> Variance.
- min()            -> Minimum value.
- max()            -> Maximum value.
- argmin()         -> Index of the minimum value.
- argmax()         -> Index of the maximum value.

#### 8. Sorting Functions
- sort()           -> Sort an array.
- argsort()        -> Return indices that would sort the array.

#### 9. Searching Functions
- where()          -> Return indices where a condition is true.
- nonzero()        -> Return indices of non-zero elements.
- unique()         -> Return unique values.

#### 10. Linear Algebra Functions
- dot()            -> Dot product.
- matmul()         -> Matrix multiplication.
- linalg.inv()     -> Matrix inverse.
- linalg.det()     -> Determinant.
- linalg.eig()     -> Eigenvalues and eigenvectors.

#### 11. Random Functions
- random.rand()        -> Random numbers between 0 and 1.
- random.randint()     -> Random integers.
- random.randn()       -> Random numbers from a normal distribution.
- random.choice()      -> Randomly choose elements.
- random.shuffle()     -> Shuffle an array.

#### 12. Logical Functions
- all()            -> Check if all elements are True.
- any()            -> Check if any element is True.
- logical_and()    -> Logical AND.
- logical_or()     -> Logical OR.
- logical_not()    -> Logical NOT.

#### 13. File Functions
- save()           -> Save an array to a file.
- load()           -> Load an array from a file.
- savetxt()        -> Save as a text file.
- loadtxt()        -> Load from a text file.

# stacking and spliting in array

the size of the array should be same 

#### horizontal stacking: row by row joining ---->np.hstack()
#### vertical stacking: col by col joining ---->np.vstack()

```python
arr1=np.arange(1,17).reshape(4,4)
arr2=np.arange(17,33).reshape(4,4)
```

```python
arr2
```

```python
total=np.hstack((arr1,arr2))
```

```python
total
```

```python
total=np.vstack((arr1,arr2))
total
```

# split funciton

#### horizontal split -----> hsplit(array,number of parts you want)
#### vertical split -----> vsplit(array,number of parts you want)


```python
split_array=np.hsplit(total,2)
split_array
```

```python
split_array=np.vsplit(total,2)
split_array
```
