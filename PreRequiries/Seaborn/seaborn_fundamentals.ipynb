# How seaborn is better than matplotlib?

#### 1. syntex --> seaborn is simpler --> less line of code is required where as in matplotlib a lot of lines of code is required for the bringing the customization 

#### 2. by default --> seaborn plotes looks classic and that also with even bringing any additional customization but in case of matplotlib by defalut its plots are simller of those of pandas plots and after adding additional customization in matplotlib we can bring those values to life 

#### 3. plot types --> matplotlib :we have general plots (line plot,scatter plots,bar,histogram,pie) but in 
####            --> seaborn : we will have all the visuals of statical analysis (Kde,Boxplot,regression plot etc) 


#### 4. integration --> matplotlib integrates with all of data type while seaborn integrate with pandas dataframe

#### 5. seaborn: more information we can extract from the plots but as compared to matplotlib 

#### 6. A lot of new plots are supported in seaborn that are not supported in matplotlib

## Figure and Axes in Matplotlib

Imagine you're drawing on paper.

- **Figure** = The entire sheet of paper.
- **Axes** = The area on the paper where you actually draw the graph.

```text
+--------------------------------------+
|              Figure                  |
|                                      |
|   +------------------------------+   |
|   |                              |   |
|   |           Axes               |   |
|   |                              |   |
|   |        Your Graph            |   |
|   |                              |   |
|   +------------------------------+   |
|                                      |
+--------------------------------------+
```

# There are 6 Classes of Plots in Seaborn

## 1. Relational Plots
- Scatter Plot (`sns.scatterplot()`)
- Line Plot (`sns.lineplot()`)

---

## 2. Distribution Plots
- Histogram (`sns.histplot()`)
- KDE Plot (`sns.kdeplot()`)
- Rug Plot (`sns.rugplot()`)

---

## 3. Categorical Plots
- Bar Plot (`sns.barplot()`)
- Count Plot (`sns.countplot()`)
- Box Plot (`sns.boxplot()`)
- Violin Plot (`sns.violinplot()`)
- Swarm Plot (`sns.swarmplot()`)

---

## 4. Matrix Plots
- Heatmap (`sns.heatmap()`)
- Cluster Map (`sns.clustermap()`)

---

## 5. Regression Plots
- Regression Plot (`sns.regplot()`)

---

## 6. Multi Plots
- Joint Plot (`sns.jointplot()`)
- Pair Plot (`sns.pairplot()`)
