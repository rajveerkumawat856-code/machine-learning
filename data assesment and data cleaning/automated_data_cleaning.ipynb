### 6.Simple Imputer ---> from Scikit learn 
- #### it is going to take the input as 2D vector   --> one col ---> impution --> 2D col  or 2D array ---> we have to convert the 2D array into 1D array by ravel() ---> simple imputer will ask out for the parameter ---> strategy ---> this decides what imputation you can go with 
- strategy --> can be mean ,median,most_frequent 


```python
df=pd.read_csv('unclean_smartwatch_health_data.csv')
```

```python
df.info()
```

```python
df=df.drop(columns='User ID')
```

```python
df.head()
```

```python
from sklearn.impute import SimpleImputer
imputer=SimpleImputer(strategy='median')
```

```python
df['Heart Rate (BPM)']=imputer.fit_transform(df[['Heart Rate (BPM)']]).ravel()
```

# we have to convert it into the 2D 

```python
df['Heart Rate (BPM)']=imputer.fit_transform(df[['Heart Rate (BPM)']]).ravel()
```

```python
df.info()
```

```python
df['Blood Oxygen Level (%)'].describe()
```

```python
imputer=SimpleImputer(strategy='mean')
```

```python
df['Blood Oxygen Level (%)']=imputer.fit_transform(df[['Blood Oxygen Level (%)']]).ravel()
```

```python
df.info()
```

```python
imputer=SimpleImputer(strategy='most_frequent')
```

```python
df['Activity Level']=imputer.fit_transform(df[['Activity Level']]).ravel()
```

```python
df.info()
```

```python
imputer=SimpleImputer(strategy='constant',fill_value=5)
df['Stress Level']=imputer.fit_transform(df[['Stress Level']]).ravel()
```
