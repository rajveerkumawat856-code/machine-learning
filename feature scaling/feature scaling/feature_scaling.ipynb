## Standardization (Z-Score Scaling)

The formula for Standardization is:

$$
X_{\text{standardized}} =
\frac{X-\mu}{\sigma}
$$

Where:

- $X$ = Original value
- $\mu$ = Mean of the feature
- $\sigma$ = Standard deviation of the feature

After standardization:

- Mean ≈ **0**
- Standard deviation ≈ **1**
- Values are **not restricted to 0–1**

Example:

Original values: 10, 20, 30, 40, 50

After standardization, values might become:

-1.41, -0.71, 0, 0.71, 1.41

| Original Value | Standardized Value |
| -------------: | -----------------: |
|             10 |              -1.41 |
|             20 |              -0.71 |
|             30 |               0.00 |
|             40 |               0.71 |
|             50 |               1.41 |


#### same for another col

## note:

## Min-Max Normalization

The formula for Min-Max Normalization is:

$$
X_{\text{normalized}} =
\frac{X - X_{\min}}
{X_{\max} - X_{\min}}
$$

Where:

- $X$ = Original value
- $X_{\min}$ = Minimum value
- $X_{\max}$ = Maximum value

The normalized values are usually between **0 and 1**.

# where the normalization is being used for ---> 
- #### neural network in deep learning  
- #### for outliears --> a huge issue will occour if data is consisting of outliears

## Why Min-Max Normalization is Sensitive to Outliers?

Min-Max Normalization uses the **minimum and maximum values**:

$$
X_{\text{normalized}} =
\frac{X-X_{\min}}
{X_{\max}-X_{\min}}
$$

If an **outlier** is present, it can make the range very large.

### Example

| Original Value | Normalized Value |
|---------------:|-----------------:|
| 10 | 0.00 |
| 20 | 0.01 |
| 30 | 0.02 |
| 40 | 0.03 |
| 50 | 0.04 |
| **1000 (Outlier)** | **1.00** |

Here:

$$
X_{\min}=10,\qquad X_{\max}=1000
$$

For $X=50$:

$$
X_{\text{normalized}}
=
\frac{50-10}{1000-10}
=
\frac{40}{990}
\approx 0.04
$$

### Why is this a problem?

The outlier **1000** stretches the range from **10 to 1000**.

As a result, the normal values **10–50 get compressed into 0–0.04**.

Therefore, Min-Max Normalization can give a poor representation when the dataset contains significant outliers.

> **Note:** Standardization is also affected by outliers because it uses the mean and standard deviation.

For datasets with significant outliers, **Robust Scaling** can be a better choice.
