# 🤖 Machine Learning

## 1. What is Machine Learning?

**Machine Learning (ML)** is a branch of Artificial Intelligence (AI) that enables computers to learn patterns from data and make predictions or decisions without being explicitly programmed for every situation.

### Simple Example

Suppose we want to predict house prices.

We provide the model with:

- Area
- Number of bedrooms
- Location
- Age of house
- House price

The ML algorithm learns the relationship between the input features and the house price. After learning, it can predict the price of a new house.

### Traditional Programming vs Machine Learning

Traditional Programming:

    Rules + Data → Program → Output

Machine Learning:

    Data + Output → ML Algorithm → Learned Pattern


# 2. Types of Machine Learning

Machine Learning is mainly divided into three types:

1. **Supervised Learning**
2. **Unsupervised Learning**
3. **Semisupervised Learning**
4. **Reinforcement Learning**



# 3. Supervised Learning

### Definition

**Supervised Learning** is a type of Machine Learning in which the model learns from labeled data, where both input features and the correct target/output are available.

### Structure

    Input Features (X) + Target (y)
                    ↓
              ML Algorithm
                    ↓
             Learned Pattern
                    ↓
              New Prediction

### Example

Suppose we have student data:

| Study Hours | Attendance | Result |
|------------|------------|--------|
| 5          | 80%        | Pass   |
| 2          | 50%        | Fail   |
| 7          | 90%        | Pass   |
| 3          | 60%        | Fail   |

Here:

- **Features (X)** = Study Hours, Attendance
- **Target (y)** = Result

The model learns from the existing data and predicts the result of a new student.

### Types of Supervised Learning

Supervised Learning is mainly divided into:

1. **Regression**
2. **Classification**


## 3.1 Regression

### Definition

**Regression** is a supervised learning technique used to predict a continuous numerical value.

### Examples

- House Price → ₹50,00,000
- Salary → ₹60,000
- Temperature → 32.5°C
- Sales → ₹1,50,000
- Number of KTs → 2

### Example

    Area + Bedrooms + Location
                ↓
        Regression Model
                ↓
          House Price

### Common Regression Algorithms

1. Linear Regression
2. Polynomial Regression
3. Ridge Regression
4. Lasso Regression
5. Elastic Net
6. KNN Regressor
7. Decision Tree Regressor
8. Random Forest Regressor
9. Gradient Boosting Regressor
10. XGBoost Regressor
11. Support Vector Regression (SVR)


## 3.2 Classification

### Definition

**Classification** is a supervised learning technique used to predict a discrete class or category.

### Examples

- Spam / Not Spam
- Pass / Fail
- Yes / No
- Fraud / Not Fraud
- Disease / No Disease
- Cat / Dog

### Example

    Study Hours + Attendance
                ↓
        Classification Model
                ↓
             Pass / Fail

### Types of Classification

#### Binary Classification

Classification with only two classes.

Examples:

- Pass / Fail
- Spam / Not Spam
- Yes / No
- Fraud / Not Fraud

#### Multiclass Classification

Classification with more than two classes.

Example:

- Low
- Medium
- High

Another example:

- Cat
- Dog
- Horse

### Common Classification Algorithms

1. Logistic Regression
2. KNN Classifier
3. Decision Tree Classifier
4. Random Forest Classifier
5. Support Vector Machine (SVM)
6. Naive Bayes
7. Gradient Boosting
8. XGBoost


# 4. Unsupervised Learning

### Definition

**Unsupervised Learning** is a type of Machine Learning in which the model learns patterns, structures, or groups from data without having a predefined target/output column.

### Structure

    Input Data (X)
          ↓
    ML Algorithm
          ↓
    Patterns / Groups

### Example

Suppose we have customer data:

| Age | Income | Spending Score |
|-----|--------|----------------|
| 20  | 20K    | 80             |
| 22  | 22K    | 85             |
| 55  | 90K    | 20             |
| 58  | 95K    | 15             |

There is no target column telling the model which customer belongs to which group.

The algorithm can discover groups based on similarities between customers.

### Main Types of Unsupervised Learning

1. **Clustering**
2. **Dimensionality Reduction**


## 4.1 Clustering

### Definition

**Clustering** is an unsupervised learning technique that groups similar data points together based on their characteristics.

### Example

    Customer Data
          ↓
       K-Means
          ↓
    ┌─────┼─────┐
    ↓     ↓     ↓
  Group  Group  Group
    1     2      3

### Applications

- Customer Segmentation
- Market Segmentation
- Document Grouping
- Image Segmentation
- Anomaly Detection

### Common Clustering Algorithms

1. K-Means
2. Hierarchical Clustering
3. DBSCAN
4. Gaussian Mixture Models


## 4.2 Dimensionality Reduction

### Definition

**Dimensionality Reduction** is the process of reducing the number of features in a dataset while retaining as much useful information as possible.

### Example

    100 Features
         ↓
        PCA
         ↓
    10 Components

### Advantages

- Reduces computational complexity
- Helps remove redundant information
- Can reduce noise
- Makes visualization easier
- Helps handle high-dimensional datasets

### Common Algorithms

1. PCA
2. t-SNE
3. UMAP


# 5. Reinforcement Learning

### Definition

**Reinforcement Learning** is a type of Machine Learning in which an agent learns to make decisions by interacting with an environment and receiving rewards or penalties for its actions.

### Basic Structure

    Agent
      ↓
    Action
      ↓
    Environment
      ↓
    Reward
      ↓
    Agent learns from the experience

### Example

A game-playing AI performs an action in a game.

- If the action is good → Positive Reward
- If the action is bad → Negative Reward

Over time, the agent learns which actions produce better rewards.

### Important Terms

- Agent
- Environment
- State
- Action
- Reward
- Policy
- Value

### Common Reinforcement Learning Algorithms

1. Q-Learning
2. SARSA
3. Deep Q-Network (DQN)
4. Policy Gradient


# 6. Comparison of Machine Learning Types

| Type | Target Available? | Main Goal | Example |
|------|--------------------|-----------|---------|
| Supervised Learning | Yes | Predict target | House Price Prediction |
| Unsupervised Learning | No | Find patterns/groups | Customer Segmentation |
| Reinforcement Learning | No fixed target | Learn through rewards | Game Playing AI |
