### Examples of Continuous Values

Continuous values are numerical values that can take any value within a range, including decimal values.

**Examples:**

* **House Price:** ₹50 lakh, ₹50.5 lakh, ₹50.55 lakh
* **Height:** 5.5 ft, 5.55 ft, 5.557 ft
* **Weight:** 70 kg, 70.2 kg, 70.25 kg
* **Temperature:** 25°C, 25.4°C, 25.43°C
* **Salary:** ₹40,000, ₹40,500.50
* **Exam Score:** 78, 78.5, 78.75

These are called continuous values because they can take many possible values, including decimal values, within a given range.
